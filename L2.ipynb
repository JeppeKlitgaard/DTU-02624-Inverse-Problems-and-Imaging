# Question 3: Abel Integration Operator

```python
N = 256
h = 1/N

s_vec = np.linspace(h/2, 1-h/2, N)
t_vec = s_vec + h/2  # Offset by h/2
t_vec = s_vec  # The same, part 3 of question 3

s_mesh, t_mesh = np.meshgrid(s_vec, t_vec)

A = np.zeros((N, N))
mask = s_mesh < t_mesh
A[mask] = h / np.sqrt(np.abs(t_mesh[mask] - s_mesh[mask]))

plt.figure()
im = plt.imshow(A)
plt.colorbar(im)
plt.title("Matrix $A$ for Abel integral operator")

U, S, V = np.linalg.svd(A)
plt.figure()
plt.semilogy(S)
plt.title('Singular values $σ$ of matrix $A$ for Abel integral operator')
plt.xlabel('Index, $j$')
plt.ylabel('Singular value, $σ_j$')
plt.tight_layout()
```
